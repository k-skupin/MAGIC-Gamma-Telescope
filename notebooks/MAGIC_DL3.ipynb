```python
from astropy.io import fits

with fits.open(file_path) as hdul:

    for hdu in hdul[1:]:

        print("\n" + "=" * 70)
        print(f"HDU: {hdu.name}")
        print("=" * 70)

        if hasattr(hdu, "columns"):
            for col in hdu.columns:
                print(
                    f"{col.name:<20} "
                    f"format={col.format:<10} "
                    f"unit={str(col.unit):<10}"
                )
```
